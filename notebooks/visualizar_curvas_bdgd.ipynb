# Validação Visual da Classificação (BDGD vs Austrália)
Este notebook varre o dataset final e plota **dinâmica e automaticamente todas as curvas** disponíveis para cada um dos grupos, sem cortes ou limites manuais.

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Carrega as curvas de telemetria agrupadas
caminho_telemetria = '../data/datasets/curvas-de-carga/Curvas_Agrupadas_Supervisionado.csv'
df_telemetria = pd.read_csv(caminho_telemetria, index_col='index')

# 2. Carrega as curvas oficiais de referência do BDGD
caminho_referencia = '../data/datasets/curvas-de-carga/CurvaCarga_referencia.csv'
df_ref = pd.read_csv(caminho_referencia, index_col='index')
```

### 📈 Comparação: `A3-Tipo01`

```python
nome_grupo = 'A3-Tipo01'
colunas_do_grupo = [col for col in df_telemetria.columns if col.startswith(f"{nome_grupo}-")]

plt.figure(figsize=(14, 5))

# Plota automaticamente TODAS as curvas que existem no dataframe para este grupo
for i, col in enumerate(colunas_do_grupo):
    label = f'Telemetria Austrália ({len(colunas_do_grupo)} plotadas)' if i == 0 else "_nolegend_"
    plt.plot(df_telemetria.index, df_telemetria[col], alpha=0.3, linewidth=1.0, color='royalblue', label=label)

# Plota a Curva Oficial do BDGD
if nome_grupo in df_ref.columns:
    plt.plot(df_ref.index, df_ref[nome_grupo], linewidth=2.5, color='crimson', linestyle='--', label=f'Referência Oficial ({nome_grupo})')

plt.title(f"Aderência das Cargas Reais à Referência: {nome_grupo}", fontsize=12, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(-0.05, 1.05)
plt.xticks([0, 144, 288, 432, 576, 720, 864], ['Seg', 'Ter', 'Qua', 'Qui', 'Sex', 'Sáb', 'Dom'])
plt.ylabel('P.U.')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
